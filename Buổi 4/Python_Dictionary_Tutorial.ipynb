# 🧠 Python Dictionary - Tổng hợp kiến thức

## 📘 1. Giới thiệu

`````python

# Dictionary (dict) là cấu trúc dữ liệu lưu trữ cặp key - value.
# Key là duy nhất, còn value thì có thể trùng.
# Tương tự như “từ điển”: mỗi từ (key) có một nghĩa (value).

example = {
    "name": "Alice",
    "age": 20,
    "is_student": True
}

print(example)
`````

`````python
dict1 = {
    "ten" : "Nguyen van hello",
    "tuoi" : 22,
    "diem" : 9.5
}
print(dict1)
`````

## 🔧 2. Tạo dictionary

`````python

# Cách 1: Dùng dấu ngoặc nhọn {}
person = {"name": "Bob", "age": 25}

# Cách 2: Dùng hàm dict()
person2 = dict(name="John", age=30)

# Cách 3: Tạo dict rỗng rồi thêm phần tử
data = {}
data["city"] = "Hanoi"
data["country"] = "Vietnam"

print(person)
print(person2)
print(data)
`````

## 🎯 3. Truy cập và sửa giá trị

`````python
# dictionary không được truy xuất qua chỉ số (index) mà thông qua khóa (key).
# Truy xuất bằng [key]: a[2], a["name"], ...
# Truy xuất bằng get(key): a.get(2), a.get("name")
# Thêm/Sửa mục: <tên_dict>[<key>] = <value>

person = {"name": "Alice", "age": 20, "city": "Hanoi"}

# Truy cập
print(person["name"])        # Alice
print(person.get("age"))     # 20
print(person.get("phone", "Không có"))  # Tránh lỗi nếu key không tồn tại

# Cập nhật
person["age"] = 21
person["email"] = "alice@gmail.com"
print(person)
`````

## 🧩 4. Xóa phần tử

`````python

student = {"name": "Nam", "grade": "A", "year": 2025}

# pop(key) - xóa và trả về giá trị
student.pop("grade")

# popitem() - xóa cặp cuối cùng (Python 3.7+)
student.popitem()

# del - xóa theo key
del student["name"]

# clear() - xóa toàn bộ
student.clear()

print(student)
`````

## 🔁 5. Duyệt qua dictionary

`````python

scores = {"Alice": 85, "Bob": 90, "Charlie": 78}

# Duyệt key
for name in scores:
    print(name)

# Duyệt value
for score in scores.values():
    print(score)

# Duyệt cả key và value
for name, score in scores.items():
    print(name, ":", score)

`````

## 🧰 6. Các phương thức hữu ích

`````python

info = {"name": "Lan", "age": 19}

# copy()
clone = info.copy()

# update() - gộp dict khác
info.update({"city": "Hue"})

# keys(), values(), items()
print(info.keys()) # trích xuất và trả về các khóa của các mục trong từ điển
print(info.values()) # trích xuất và trả về các giá trị của các mục trong từ điển
print(info.items()) # trích xuất và trả về các cặp (key, value) của các mục trong từ điển

# fromkeys() - tạo dict từ danh sách key
# dict.fromkeys(alphabets,number): tạo một từ điển từ chuỗi khóa và giá trị  đã cho
# alphabets – dữ liệu các khóa được lưu trữ bởi chuỗi, tập hợp, danh sách,...
# numbers (tùy chọn) – là giá trị thuộc bất kỳ kiểu dữ liệu nào.
# Giá trị giống nhau được gán cho tất cả các khóa của từ điển
# Mặc định giá trị None được gán cho các khóa của từ điển nếu không có giá  trị được truyền vào.

keys = ["a", "b", "c"]
new_dict = dict.fromkeys(keys, 0)
print(new_dict)

`````

## 💡 7. Dictionary comprehension

`````python

# Tạo dict từ comprehension
# { <biểu_thức_key> : <biểu_thức_value>  for <biến> in <iterable>  if <điều_kiện_tùy_chọn> }

squares = {x: x**2 for x in range(5)}
print(squares)

# Lọc phần tử
even_squares = {x: x**2 for x in range(10) if x % 2 == 0}
print(even_squares)

`````

## 🧮 8. Bài tập thực hành

Trong Python, hàm `sorted()` dùng để **sắp xếp dữ liệu** và **trả về một danh sách mới đã sắp xếp**, không làm thay đổi dữ liệu ban đầu.

#### **Cú pháp cơ bản**

```python
sorted(iterable, key=None, reverse=False)
````

| Tham số    | Ý nghĩa                                                  |
| ---------- | -------------------------------------------------------- |
| `iterable` | Dữ liệu có thể duyệt được (list, tuple, set, dict, v.v.) |
| `key`      | Hàm xác định tiêu chí so sánh (thường dùng `lambda`)     |
| `reverse`  | `False` → tăng dần (mặc định), `True` → giảm dần         |

---

### 🔥 `lambda` trong `sorted()`
🔍 Hiểu rõ lambda trong sorted()

key=lambda x: <biểu_thức>

x đại diện cho mỗi phần tử trong dữ liệu.

lambda trích ra khóa để Python lấy làm tiêu chí sắp xếp.

`lambda` giúp chỉ rõ **sắp xếp theo giá trị nào**.

#### 1) Sắp xếp **đối với dict.items()**

Khi duyệt một dictionary bằng `.items()`, mỗi phần tử là một tuple dạng `(key, value)`:

```python
(name, score)
```

Muốn sắp xếp theo **giá trị** (`score`), ta dùng:

```python
sorted(scores.items(), key=lambda x: x[1])
```

* `x` là từng phần tử dạng `(name, score)`
* `x[1]` là phần **giá trị** dùng để so sánh

#### Nếu muốn sắp xếp **giảm dần**, thêm `reverse=True`:

```python
sorted(scores.items(), key=lambda x: x[1], reverse=True)
```
---

### Tóm tắt

| Mục tiêu                          | Câu lệnh                                                   |
| --------------------------------- | ---------------------------------------------------------- |
| Sắp xếp **tăng dần theo giá trị** | `sorted(scores.items(), key=lambda x: x[1])`               |
| Sắp xếp **giảm dần theo giá trị** | `sorted(scores.items(), key=lambda x: x[1], reverse=True)` |
| Chuyển về `dict` sau sắp xếp      | `dict(sorted(...))`                                        |
```


`````python
# 🧩 BÀI 1 — Quản lý điểm học sinh
# -------------------------------------------------------------
# Yêu cầu:
# 1. Cho danh sách điểm học sinh dưới dạng dict:
#    {"An": 8.5, "Bình": 7.2, "Chi": 9.0, "Dũng": 6.8, "Hà": 8.0}
# 2. Viết chương trình thực hiện các yêu cầu sau:
#    - In danh sách học sinh và điểm của họ.
#    - Tìm học sinh có điểm cao nhất và thấp nhất.
#    - Tính điểm trung bình của cả lớp.
#    - Tạo một dict mới lưu xếp loại (>=8: "Giỏi", 6.5–7.9: "Khá", <6.5: "Trung bình").
#    - Sắp xếp danh sách học sinh theo điểm giảm dần.
# -------------------------------------------------------------
`````

`````python
danhsach = {
    "An": 8.5,
    "Bình": 7.2,
    "Chi": 9.0,
    "Dũng": 6.8,
    "Hà": 8.0
}
print(danhsach)
for ten, diem in danhsach.items():
    if diem == max(danhsach.values()):
        print("Học sinh có điểm cao nhất:", ten, diem)
    if diem == min(danhsach.values()):
        print("Học sinh có điểm thấp nhất:", ten, diem)
# Tính điểm trung bình
tong_diem = sum(danhsach.values())
TBC = tong_diem / len(danhsach)
print("Điểm trung bình của cả lớp:", TBC)
# dict moi luu xep loai


`````

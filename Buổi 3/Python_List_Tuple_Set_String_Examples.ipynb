🧩 Bài 1 – Danh sách điểm sinh viên (List)

Yêu cầu:
Viết chương trình quản lý danh sách điểm của sinh viên:

Nhập danh sách điểm từ bàn phím (các số cách nhau bởi dấu cách).

Loại bỏ điểm nhỏ hơn 5.

Sắp xếp danh sách còn lại theo thứ tự giảm dần.

In ra danh sách điểm hợp lệ và điểm cao nhất, thấp nhất.

```python
ds = [6,7,8,10,3,2,4,5.4,2.7]
ds1 = [x for x in ds if x >= 5]
ds1.sort(reverse=True)
print("Danh sách điểm hợp lệ (≥5):", ds1)
if ds1:
    print("Điểm cao nhất:", max(ds1))
    print("Điểm thấp nhất:", min(ds1))
else:
    print("Không có điểm hợp lệ (≥5).")
```

### 🔹 F-string & format

```python

name = "Quyen"
age = 20
print(f"My name is {name}, I'm {age} years old.")
print("My name is {}, I'm {} years old.".format(name, age))
```

```python
s = "Python cơ bản"

print(s[0:6])   # "Python" – từ index 0 đến 5
print(s[:6])    # giống trên
print(s[7:])    # "cơ bản" – từ index 7 đến hết
print(s[:])     # sao chép toàn bộ chuỗi
```

```python
text = "abcdefg"

print(text[::2])   # "aceg" – lấy cách 2 ký tự
print(text[::-1])  # "gfedcba" – đảo ngược chuỗi
print(text[1:6:2]) # "bdf"
```

⚠️ Một số cú pháp không hợp lệ:
Cú pháp	Lỗi hoặc kết quả
s[2:1]	"" (vì start > end với step dương)
s[::0]	❌ ValueError: slice step cannot be zero
s[100:]	"" (vượt giới hạn sẽ không lỗi, chỉ trả chuỗi rỗng)

```python
palindrome = "madam"
print(palindrome == palindrome[::-1])  # True
```

```python
filename = "report.pdf"
print(filename[-3:])  # "pdf"
```

🧩 Bài 2 – Xử lý chuỗi (String)

Yêu cầu:
Viết chương trình nhận vào một chuỗi, sau đó:

Đếm số lượng nguyên âm và phụ âm trong chuỗi.

In ra chuỗi đảo ngược.

Kiểm tra xem chuỗi có phải là palindrome (đọc xuôi ngược như nhau) không.

```python
chuoi = input("Nhap chuoi ki tu: ")
nguyen_am = "u e o a i  U E O A I "
dem = 0
dem1 =0
for i in chuoi:
    if i in nguyen_am:
        dem +=1
    else:
        dem1 +=1
print("So nguyen am:", dem)
print("So phu am:", dem1)
print("Chuoi dao nguoc: ",chuoi[::-1])
dao_nguoc = chuoi[::-1]
if chuoi == dao_nguoc:
    print("la palindrome")
else:
    print ("ko la palindrome")
```
